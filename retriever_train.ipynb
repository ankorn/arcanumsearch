```python
from datasets import load_dataset
from sentence_transformers import (
    SentenceTransformer,
    SentenceTransformerTrainer,
    SentenceTransformerTrainingArguments,
)
from sentence_transformers.losses import GISTEmbedLoss
from sentence_transformers.training_args import BatchSamplers
from sentence_transformers.evaluation import InformationRetrievalEvaluator
from sentence_transformers.util import mine_hard_negatives
import torch
```

```python
dataset = load_dataset("pameydorke/arcanumchat-cross-platform-with-vague-queries-qwen-36-a3b-fp8", download_mode="force_redownload")

train_dataset = dataset["train"]
val_dataset = dataset["validation"]
```

```python
model = SentenceTransformer('BAAI/bge-m3')

guide_model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

loss = GISTEmbedLoss(
    model=model,
    guide=guide_model,
    temperature=0.07
)
```

```python
queries = {}
corpus = {}
relevant_docs = {}

unique_docs = {}
for idx, example in enumerate(val_dataset):
    doc_text = example['text']
    if doc_text not in unique_docs:
        doc_id = example['doc_id']
        unique_docs[doc_text] = doc_id
        corpus[doc_id] = doc_text

for idx, example in enumerate(val_dataset):
    query = example['query']
    doc_text = example['text']
    doc_id = unique_docs[doc_text]
    query_id = f"query_{idx}"
    queries[query_id] = query
    relevant_docs[query_id] = {doc_id}

evaluator = InformationRetrievalEvaluator(
    queries=queries,
    corpus=corpus,
    relevant_docs=relevant_docs,
    name="val-hard-negatives",
    show_progress_bar=False,
)
```

```python
train_dataset = train_dataset.select_columns(["query", "text"])
train_dataset = train_dataset.rename_columns({
    "query": "anchor",
    "text": "positive"
})

val_dataset = val_dataset.select_columns(["query", "text"])
val_dataset = val_dataset.rename_columns({
    "query": "anchor",
    "text": "positive"
})
```

```python
hard_eval_dataset = mine_hard_negatives(
    dataset=val_dataset,
    model=model,
    corpus=list(val_dataset["positive"]),

    num_negatives=4,

    range_min=1,
    range_max=100,
    relative_margin=0.05,
    sampling_strategy="top",
    batch_size=128,
    
    use_faiss=True,
)
```

```python
hard_train_dataset = mine_hard_negatives(
    dataset=train_dataset,
    model=model,
    corpus=list(train_dataset["positive"]),

    num_negatives=4,

    range_min=1,
    range_max=100,
    relative_margin=0.05,
    sampling_strategy="top",
    batch_size=128,
    
    use_faiss=True,
)
```

```python
training_args = SentenceTransformerTrainingArguments(
    warmup_steps=0.1,
    lr_scheduler_type="linear",

    num_train_epochs=1,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    gradient_accumulation_steps = 2,

    logging_steps=100,

    eval_strategy="steps",
    eval_steps=100,
    save_strategy="steps",
    save_steps=100,
    save_total_limit=2,

    load_best_model_at_end=True,
    metric_for_best_model="eval_val-hard-negatives_cosine_recall@10",

    greater_is_better=True,

    batch_sampler=BatchSamplers.NO_DUPLICATES, # set if embed loss

    eval_on_start=True,
)
```

```python
from transformers import EarlyStoppingCallback

trainer = SentenceTransformerTrainer(
    model=model,
    args=training_args,
    train_dataset=hard_train_dataset,
    eval_dataset=hard_eval_dataset,
    loss=loss,
    evaluator=evaluator,
    # callbacks=[
    #         EarlyStoppingCallback(
    #             early_stopping_patience=5,
    #             early_stopping_threshold=0.02,
    #         )
    #     ],
)

trainer.train()
```
